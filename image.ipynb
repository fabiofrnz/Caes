```python
from PIL import Image

def process_image(image_path, output_gray_path, output_binary_path, threshold=128):
    """
    Converte uma imagem colorida para níveis de cinza e binarizada (preto e branco).

    :param image_path: Caminho da imagem de entrada (colorida).
    :param output_gray_path: Caminho para salvar a imagem em tons de cinza.
    :param output_binary_path: Caminho para salvar a imagem binarizada.
    :param threshold: Valor limite para binarização (padrão: 128).
    """
    try:
        # Abrir a imagem colorida
        image = Image.open(image_path)

        # Converter para tons de cinza
        gray_image = image.convert("L")
        gray_image.save(output_gray_path)

        # Converter para imagem binarizada
        binary_image = gray_image.point(lambda x: 255 if x >= threshold else 0, '1')
        binary_image.save(output_binary_path)

        print(f"Imagens processadas com sucesso!\nTons de cinza: {output_gray_path}\nBinarizada: {output_binary_path}")

    except Exception as e:
        print(f"Erro ao processar a imagem: {e}")

# Exemplo de uso
# Substitua os caminhos pelos de sua preferência.
process_image(
    image_path="entrada.jpg",
    output_gray_path="saida_cinza.jpg",
    output_binary_path="saida_binaria.jpg",
    threshold=128
)
```
